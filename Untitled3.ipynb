```python
# CELL 2: Select and upload the original dataset directly

import pandas as pd
from pathlib import Path
import tkinter as tk
from tkinter import filedialog

# Open a file selection window
root = tk.Tk()
root.withdraw()
root.attributes("-topmost", True)

file_path = filedialog.askopenfilename(
    title="Select Your Original Dataset",
    filetypes=[
        ("Supported Dataset Files", "*.csv *.xlsx *.xls"),
        ("CSV Files", "*.csv"),
        ("Excel Files", "*.xlsx *.xls"),
        ("All Files", "*.*")
    ]
)

root.destroy()

# Check file selection
if not file_path:
    print("No file selected. Please run this cell again.")
else:
    path = Path(file_path)
    extension = path.suffix.lower()

    # Load the selected dataset
    if extension == ".csv":
        df = pd.read_csv(path, encoding="utf-8-sig")
    elif extension == ".xlsx":
        df = pd.read_excel(path, engine="openpyxl")
    elif extension == ".xls":
        df = pd.read_excel(path, engine="xlrd")
    else:
        raise ValueError("Please select a CSV, XLSX, or XLS file.")

    if df.empty:
        raise ValueError("The selected dataset is empty.")

    print("Original dataset uploaded successfully!")
    print("Selected file:", path.name)
    print("Dataset shape:", df.shape)
    print("\nFirst five rows:")

    display(df.head())
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst five rows:")
display(df.head())

print("\nLast five rows:")
display(df.tail())
```

```python
df.columns = df.columns.astype(str).str.strip()

df = df.loc[
    :,
    ~df.columns.str.match(r"^Unnamed", case=False)
].copy()

print("Unnecessary columns removed successfully!")
print(df.columns.tolist())
```

```python
print("Duplicate rows before cleaning:")
print(df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed successfully!")
print("Dataset shape after duplicate removal:", df.shape)
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
text_columns = ["Player", "Span"]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype("string").str.strip()

print("Text columns cleaned successfully!")
display(df[text_columns].head())
```

```python
if "HS" in df.columns:
    hs_text = df["HS"].astype("string").str.strip()

    if "HS_NotOut" not in df.columns:
        df["HS_NotOut"] = hs_text.str.contains(
            r"\*", na=False
        )
    else:
        # Preserve an existing not-out indicator and
        # recognize any asterisk in the highest score.
        existing = df["HS_NotOut"].astype("string").str.lower()
        existing_bool = existing.isin(
            ["true", "1", "yes", "y"]
        )

        df["HS_NotOut"] = (
            existing_bool
            | hs_text.str.contains(r"\*", na=False)
        )

    # Remove the asterisk from the numeric score.
    df["HS"] = hs_text.str.replace(
        "*", "", regex=False
    ).str.strip()

print("Highest score and not-out indicator processed!")
display(df[["HS", "HS_NotOut"]].head())
```

```python
numeric_columns = [
    "Mat", "Inns", "NO", "Runs", "HS",
    "Ave", "BF", "SR", "100", "50", "0",
    "4s", "6s"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
            .str.replace(",", "", regex=False)
            .str.replace("+", "", regex=False)
        )

        df[col] = pd.to_numeric(
            df[col], errors="coerce"
        )

print("Invalid numeric symbols handled successfully!")
print(df[numeric_columns].dtypes)
```

```python
for col in numeric_columns:
    if col in df.columns:
        df[col] = df[col].fillna(0)

for col in ["Player", "Span"]:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

if "HS_NotOut" in df.columns:
    df["HS_NotOut"] = df["HS_NotOut"].fillna(False)

print("Missing values handled successfully!")

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nTotal missing values remaining:")
print(df.isnull().sum().sum())
```

```python
integer_columns = [
    "Mat", "Inns", "NO", "Runs", "HS",
    "BF", "100", "50", "0", "4s", "6s"
]

float_columns = ["Ave", "SR"]

for col in integer_columns:
    if col in df.columns:
        df[col] = df[col].round().astype("int64")

for col in float_columns:
    if col in df.columns:
        df[col] = df[col].astype("float64")

if "HS_NotOut" in df.columns:
    df["HS_NotOut"] = df["HS_NotOut"].astype(bool)

print("Data types corrected successfully!")

print("\nFinal data types:")
print(df.dtypes)
```

```python
print("Checking the cleaned dataset...")

print("\nNegative numerical values:")
for col in numeric_columns:
    if col in df.columns:
        print(f"{col}: {(df[col] < 0).sum()}")

print("\nDuplicate rows remaining:")
print(df.duplicated().sum())

print("\nMissing values remaining:")
print(df.isnull().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
required_columns = [
    "Player", "Span", "Mat", "Inns", "NO",
    "Runs", "HS", "Ave", "BF", "SR",
    "100", "50", "0", "4s", "6s",
    "HS_NotOut"
]

# Retain the expected column order when all columns exist.
if all(col in df.columns for col in required_columns):
    df = df[required_columns].copy()

print("Column order checked successfully!")
print(df.columns.tolist())
```
