```python
import pandas as pd

df = pd.read_excel("odb.xlsx", sheet_name="odb (1)")
df.head()
```

```python
df = df.drop(columns=["Unnamed: 0"])

df = df.rename(columns={
    "Mat": "Matches", "Inns": "Innings", "NO": "Not_Outs",
    "HS": "High_Score", "Ave": "Average", "BF": "Balls_Faced",
    "SR": "Strike_Rate", "100": "Hundreds", "50": "Fifties",
    "0": "Ducks", "4s": "Fours", "6s": "Sixes"
})
df.head()
```

```python
df["Not_Out_HS"] = df["High_Score"].astype(str).str.contains(r"\*")
df["High_Score"] = df["High_Score"].astype(str).str.replace("*", "", regex=False).astype(int)

for col in ["Fours", "Sixes"]:
    df[col] = df[col].astype(str).str.replace("+", "", regex=False).astype(int)

df.info()
```

```python
df["Country_Raw"] = df["Player"].str.extract(r"\((.*)\)")
df["Player_Name"] = df["Player"].str.replace(r"\s*\(.*\)", "", regex=True).str.strip()

regions = {"Asia", "Afr", "ICC"}
def main_country(s):
    parts = [p for p in s.split("/") if p not in regions]
    return parts[-1] if parts else s

df["Country"] = df["Country_Raw"].apply(main_country)

df[["Player", "Player_Name", "Country_Raw", "Country"]].head(10)
```

```python
df[["Start_Year", "End_Year"]] = df["Span"].str.split("-", expand=True).astype(int)
df["Career_Years"] = df["End_Year"] - df["Start_Year"]

df[["Player_Name", "Span", "Start_Year", "End_Year", "Career_Years"]].head(10)
```
